1. Lectura de un archivo HTML

En este bloque se ve como Python puede leer y analizar un archivo HTML de la interfaz de Matriculator.

```python
html = """
<html>
<head>
    <title>Matriculator</title>
</head>
<body>
    <h1>Matriculator</h1>
    <p>Deteccion de matriculas en imagenes de prueba</p>

    <ul>
        <li>Subir imagen</li>
        <li>Detectar matricula</li>
        <li>Mostrar resultado</li>
    </ul>
</body>
</html>
"""

with open("interfaz.html", "w") as archivo:
    archivo.write(html)
```

```python
from bs4 import BeautifulSoup

with open("interfaz.html", "r") as archivo:
    contenido = archivo.read()

soup = BeautifulSoup(contenido, "html.parser")

lista = soup.find_all("li")

for elemento in lista:
    print(elemento.text)
```

3. Lectura de un archivo JSON

En este bloque se muestra como Python puede leer los nuevos datos de entrada sinteticos que llegarian a la aplicacion en formato JSON (la imagen subida y sus datos de contexto).

```python
import json

# Datos de entrada sinteticos: asi llegaria la informacion de cada imagen a la aplicacion
datos = {
    "imagen": "entrada_01.jpg",
    "formato": "JPG",
    "tamano_mb": 2.4,
    "camara": "entrada_norte",
    "fecha": "2026-09-30 08:15:00",
    "sintetica": True
}

# Se guarda en un fichero JSON
with open("entradas.json", "w") as archivo:
    json.dump(datos, archivo, indent=4)

# Se vuelve a leer: el texto del fichero se convierte en un diccionario de Python
with open("entradas.json", "r") as archivo:
    datos_leidos = json.load(archivo)

print(datos_leidos)

# Se accede a cada campo por su clave
print("Imagen:", datos_leidos["imagen"])
print("Camara:", datos_leidos["camara"])
print("Tamano (MB):", datos_leidos["tamano_mb"])
```

Librerias principales de Python

- HTML: BeautifulSoup (bs4), que analiza el codigo HTML y permite buscar etiquetas como <li>.
- CSV: pandas, que lee el fichero como una tabla (DataFrame) con read_csv.
- JSON: json, incluida en Python, que convierte el JSON en diccionarios y listas.

Librerias equivalentes en R

| Tipo de archivo | Python | R |
|---|---|---|

| CSV | pandas | readr |
